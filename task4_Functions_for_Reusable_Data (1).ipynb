```python
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

marks = [80, 75, 90, 85, 70]

result = calculate_mean(marks)

print("Mean Marks:", result)
```

```python
def calculate_maximum(numbers):
    return max(numbers)

marks = [80, 75, 90, 85, 70]

result = calculate_maximum(marks)

print("Maximum Marks:", result)
```

```python
def calculate_minimum(numbers):
    return min(numbers)

marks = [80, 75, 90, 85, 70]

result = calculate_minimum(marks)

print("Minimum Marks:", result)
```

```python
def calculate_percentage(obtained_marks, total_marks):
    if total_marks <= 0:
        return None

    return (obtained_marks / total_marks) * 100

percentage = calculate_percentage(420, 500)

print("Percentage:", percentage, "%")
```

```python
def calculate_grade(percentage):
    if percentage is None or percentage < 0 or percentage > 100:
        return "Invalid Percentage"
    elif percentage >= 90:
        return "A"
    elif percentage >= 80:
        return "B"
    elif percentage >= 70:
        return "C"
    elif percentage >= 60:
        return "D"
    else:
        return "F"

grade = calculate_grade(84)

print("Grade:", grade)
```
